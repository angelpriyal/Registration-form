```python

# 1. Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
airline_df = pd.read_csv("airline_dec_2008_50k.csv")
```

```python
# 3. Display first five records
print(airline_df.head())

# 4. Display dataset information
print(airline_df.info())

# 5. Select the Distance column
distance = airline_df['Distance']

# 6. Calculate descriptive statistics
mean_distance = distance.mean()
median_distance = distance.median()
skewness_distance = distance.skew()
kurtosis_distance = distance.kurt()
```

```python

# 7. Print calculated values
print("Mean:", mean_distance)
print("Median:", median_distance)
print("Skewness:", skewness_distance)
print("Kurtosis:", kurtosis_distance)
```

```python
# 8. Plot histogram
plt.figure(figsize=(8, 6))

sns.histplot(distance, kde=True, color='skyblue', bins=30)

# 9. Display mean and median
plt.axvline(mean_distance, color='red',
            linestyle='--',
            label=f'Mean: {mean_distance:.2f}')

plt.axvline(median_distance, color='black',
            linestyle='--',
            label=f'Median: {median_distance:.2f}')

# 10. Add title and labels
plt.title(
    f'Distribution of Distance\n'
    f'Skewness: {skewness_distance:.2f}, '
    f'Kurtosis: {kurtosis_distance:.2f}'
)


plt.xlabel('Distance')
plt.ylabel('Frequency')
```
